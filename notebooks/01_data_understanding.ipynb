### Variant ID Relationship

The `variant_id` field in `orders_items` corresponds to the `id` field in `products_skus`.

A sample lookup confirmed that the same variant ID exists in both tables and is associated with the same SKU and product style.

Therefore, the intended relationship is:

`orders_items.variant_id` → `products_skus.id`


```python
# 匹配不上的订单商品记录数量
unmatched_variant_row_count = (
    ~orders_items["variant_id"].isin(products_skus["id"])
).sum()

# 匹配不上的不同 variant_id 数量
unmatched_variant_id_count = (
    orders_items.loc[
        ~orders_items["variant_id"].isin(products_skus["id"]),
        "variant_id"
    ]
    .nunique()
)

print(
    "Order-item rows with variant_id not found:",
    unmatched_variant_row_count
)

print(
    "Unique variant IDs not found:",
    unmatched_variant_id_count
)
```

```python
# 匹配不上的订单商品记录数量
unmatched_sku_row_count = (
    ~orders_items["sku"].isin(products_skus["sku"])
).sum()

# 匹配不上的不同 SKU 数量
unmatched_sku_count = (
    orders_items.loc[
        ~orders_items["sku"].isin(products_skus["sku"]),
        "sku"
    ]
    .nunique()
)

print(
    "Order-item rows with SKU not found:",
    unmatched_sku_row_count
)

print(
    "Unique SKUs not found:",
    unmatched_sku_count
)
```
